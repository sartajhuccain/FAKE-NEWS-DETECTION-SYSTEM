```python
import os
import glob
import re
import warnings
import random

import kagglehub
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

warnings.filterwarnings("ignore")

SEED = 42
random.seed(SEED)
np.random.seed(SEED)

print("All libraries imported successfully!")
```

```python
# Download dataset
path = kagglehub.dataset_download(
    "emineyetm/fake-news-detection-datasets"
)

print("Dataset downloaded successfully!")
print("Dataset path:", path)
```

```python
# Find all CSV files
csv_files = glob.glob(
    os.path.join(path, "**", "*.csv"),
    recursive=True
)

print("Number of CSV files:", len(csv_files))

for file in csv_files:
    print(file)

if len(csv_files) == 0:
    raise FileNotFoundError(
        "No CSV files found. Check the downloaded dataset folder."
    )
```

```python
def find_csv_by_name(files, target_name):
    target_name = target_name.lower()

    matches = [
        f for f in files
        if os.path.basename(f).lower() == target_name
    ]

    if not matches:
        return None

    return matches[0]


fake_path = find_csv_by_name(csv_files, "Fake.csv")
true_path = find_csv_by_name(csv_files, "True.csv")

if fake_path and true_path:

    fake_df = pd.read_csv(fake_path, low_memory=False)
    real_df = pd.read_csv(true_path, low_memory=False)

    fake_df["label"] = 0
    real_df["label"] = 1

    df = pd.concat(
        [fake_df, real_df],
        ignore_index=True
    )

else:
    print("Fake.csv or True.csv was not found.")
    print("Available CSV files:")

    for file in csv_files:
        print(os.path.basename(file))

    raise FileNotFoundError(
        "Please inspect the files above and adjust the loading code."
    )

# Standardize column names
df.columns = [
    str(col).strip().lower()
    for col in df.columns
]

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

display(df.head())
```

```python
print("Dataset Information")
print("=" * 50)

print("Total articles:", len(df))

print("\nColumn names:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum())

print("\nLabel distribution:")
print(df["label"].value_counts())

print("\nLabel percentages:")
print(
    (df["label"].value_counts(normalize=True) * 100)
    .round(2)
)
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="label"
)

plt.title("Fake News vs Real News")
plt.xlabel("Label: 0 = Fake, 1 = Real")
plt.ylabel("Number of Articles")

plt.show()
```

```python
# Ensure the expected text columns exist
if "text" not in df.columns:
    raise ValueError(
        f"'text' column not found. Available columns: {df.columns.tolist()}"
    )

if "title" not in df.columns:
    df["title"] = ""

# Handle missing values
df["title"] = df["title"].fillna("").astype(str)
df["text"] = df["text"].fillna("").astype(str)

# Combine title and article body
df["news"] = (
    df["title"].str.strip()
    + " "
    + df["text"].str.strip()
).str.strip()

# Remove rows with no meaningful text
df = df[df["news"].str.len() > 0].copy()

# Remove duplicate articles
df = df.drop_duplicates(subset=["news"])

# Keep only valid labels
df = df[df["label"].isin([0, 1])].copy()
df["label"] = df["label"].astype(int)

df = df.reset_index(drop=True)

print("Cleaning completed!")
print("Remaining articles:", len(df))

display(df[["title", "text", "news", "label"]].head())
```

```python
def clean_text(text):
    text = str(text).lower()

    # Remove HTML tags
    text = re.sub(r"<[^>]+>", " ", text)

    # Normalize URLs and email addresses
    text = re.sub(r"https?://\S+|www\.\S+", " URL ", text)
    text = re.sub(r"\S+@\S+", " EMAIL ", text)

    # Remove repeated whitespace
    text = re.sub(r"\s+", " ", text)

    return text.strip()


df["news"] = df["news"].apply(clean_text)

print("Text preprocessing completed!")
print(df["news"].iloc[0][:500])
```

```python
X = df["news"]
y = df["label"]

print("Feature shape:", X.shape)
print("Label shape:", y.shape)

print("\nFake articles:", (y == 0).sum())
print("Real articles:", (y == 1).sum())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=SEED,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

print(
    "Training percentage:",
    round(len(X_train) / len(X) * 100, 2),
    "%"
)

print(
    "Testing percentage:",
    round(len(X_test) / len(X) * 100, 2),
    "%"
)
```

```python
tfidf = TfidfVectorizer(
    stop_words="english",
    max_features=100000,
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.95,
    sublinear_tf=True,
    dtype=np.float32
)

X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

print("TF-IDF conversion completed!")

print("Training matrix shape:", X_train_tfidf.shape)
print("Testing matrix shape:", X_test_tfidf.shape)
```

```python
lr_model = LogisticRegression(
    max_iter=1000,
    C=2.0,
    random_state=SEED
)

print("Training Logistic Regression...")

lr_model.fit(
    X_train_tfidf,
    y_train
)

print("Logistic Regression training completed!")
```

```python
lr_predictions = lr_model.predict(X_test_tfidf)

lr_accuracy = accuracy_score(
    y_test,
    lr_predictions
)

print(
    f"Logistic Regression Accuracy: {lr_accuracy * 100:.2f}%"
)

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        lr_predictions,
        target_names=["Fake News", "Real News"],
        zero_division=0
    )
)
```

```python
svm_model = LinearSVC(
    C=1.0,
    random_state=SEED,
    max_iter=5000
)

print("Training Linear SVM...")

svm_model.fit(
    X_train_tfidf,
    y_train
)

print("Linear SVM training completed!")
```

```python
svm_predictions = svm_model.predict(X_test_tfidf)

svm_accuracy = accuracy_score(
    y_test,
    svm_predictions
)

print(
    f"Linear SVM Accuracy: {svm_accuracy * 100:.2f}%"
)

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        svm_predictions,
        target_names=["Fake News", "Real News"],
        zero_division=0
    )
)
```

```python
results = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Linear SVM"
    ],
    "Accuracy (%)": [
        lr_accuracy * 100,
        svm_accuracy * 100
    ]
})

results = results.sort_values(
    by="Accuracy (%)",
    ascending=False
).reset_index(drop=True)

display(results)

plt.figure(figsize=(8, 5))

sns.barplot(
    data=results,
    x="Model",
    y="Accuracy (%)"
)

plt.title("Fake News Detection Model Comparison")
plt.ylim(0, 100)
plt.ylabel("Accuracy (%)")

plt.show()
```

```python
best_model_name = (
    "Logistic Regression"
    if lr_accuracy >= svm_accuracy
    else "Linear SVM"
)

best_model = (
    lr_model
    if lr_accuracy >= svm_accuracy
    else svm_model
)

best_predictions = best_model.predict(X_test_tfidf)

print("Selected model:", best_model_name)

cm = confusion_matrix(
    y_test,
    best_predictions,
    labels=[0, 1]
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Fake News", "Real News"]
)

disp.plot(
    cmap="Blues",
    values_format="d"
)

plt.title("Confusion Matrix")
plt.show()
```

```python
best_pipeline = Pipeline([
    ("tfidf", tfidf),
    ("classifier", best_model)
])

MODEL_PATH = "/content/fake_news_detection_model.joblib"

joblib.dump(
    best_pipeline,
    MODEL_PATH
)

print("Model saved successfully!")
print("Saved model path:", MODEL_PATH)
```

```python
def predict_news(title, article_text):
    combined_news = str(title) + " " + str(article_text)
    combined_news = clean_text(combined_news)

    if not combined_news.strip():
        raise ValueError(
            "Please provide a news headline or article text."
        )

    prediction = loaded_model.predict(
        [combined_news]
    )[0]

    if prediction == 0:
        result = "FAKE NEWS"
    else:
        result = "REAL NEWS"

    print("Prediction:", result)
    print(
        "Note: This is a model prediction, "
        "not a verified fact-check."
    )

    return result
```
